Дорога между A и B состоит из подъёма и спуска. Велосипедист, двигаясь на спуске со скоростью 18 км/ч, а на
подъёме 12 км/ч, затрачивает на путь от A до B 2 ч. 40 мин. Найти (в км.) длину подъёма в направлении от A до B,
если длина всей дороги 36 км.   

```python
from sympy.abc import x
from sympy import Eq
from sympy import solve

sol = solve(Eq(x/12+(36-x)/18,2+2/3), [x])
print("CB=", sol[0], " km", sep="")
```

```python
from sympy.abc import x, y
from sympy import Eq
from sympy import solve

sol = solve([Eq(x + y, 36), Eq(x / 12 + y / 18, 2+2/3)], [x,y])
print("CB=", sol[x], " km", sep="")
```
